```python
print(
    "Points with invalid case_id:",
    (~points["case_id"].isin(cases["id"])).sum()
)

print(
    "Points with invalid service_id:",
    (~points["service_id"].isin(services["id"])).sum()
)

print(
    "Cases with invalid topic_id:",
    (~cases["topic_id"].isin(topics["id"])).sum()
)
```

```python
approved_points = points[points["status"] == "approved"]

print("Approved points:", len(approved_points))

print(
    "Approved points with missing quote_text:",
    approved_points["quote_text"].isna().sum()
)

print(
    "Approved points with valid quote_text:",
    approved_points["quote_text"].notna().sum()
)
```

```python
points_cases = points.merge(
    cases[["id", "classification", "topic_id"]],
    left_on="case_id",
    right_on="id",
    how="left"
)

print(
    pd.crosstab(
        points_cases["status"],
        points_cases["classification"]
    )
)
```
